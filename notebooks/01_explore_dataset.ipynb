```python
uber_samples = []

for chunk in pd.read_csv("../data/twcs.csv", chunksize=100_000):
    
    matches = chunk[
        (chunk["author_id"] == "Uber_Support") &
        (chunk["inbound"] == False)
    ]
    
    if len(matches) > 0:
        uber_samples.append(matches)

uber_df = pd.concat(uber_samples, ignore_index=True)

print("Uber brand responses:", len(uber_df))
```

```python
# Customer messages used in the golden evaluation set
golden_customers = set(golden_set["text_customer"])

# Remove golden examples from the historical retrieval data
retrieval_corpus = uber_pairs[
    ~uber_pairs["text_customer"].isin(golden_customers)
].copy()

print("Original pairs:", len(uber_pairs))
print("Retrieval corpus:", len(retrieval_corpus))
print("Golden examples excluded:", len(uber_pairs) - len(retrieval_corpus))
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

retriever_vectorizer = TfidfVectorizer(
    lowercase=True,
    stop_words="english",
    ngram_range=(1, 2),
    max_features=50000
)

retriever_matrix = retriever_vectorizer.fit_transform(
    retrieval_corpus["text_customer"]
)

print("Retrieval matrix shape:", retriever_matrix.shape)
```

```python
def retrieve_similar_cases(query, top_k=3):

    query_vector = retriever_vectorizer.transform([query])

    similarities = cosine_similarity(
        query_vector,
        retriever_matrix
    ).flatten()

    top_indices = similarities.argsort()[-top_k:][::-1]

    results = retrieval_corpus.iloc[top_indices].copy()
    results["similarity"] = similarities[top_indices]

    return results
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# Exclude golden-set customer messages to prevent evaluation leakage
golden_customers = set(golden_eval["text_customer"])

retrieval_corpus = uber_pairs[
    ~uber_pairs["text_customer"].isin(golden_customers)
].copy()

print("Original pairs:", len(uber_pairs))
print("Retrieval corpus:", len(retrieval_corpus))
print("Golden examples excluded:",
      len(uber_pairs) - len(retrieval_corpus))


# Build TF-IDF representation
retriever_vectorizer = TfidfVectorizer(
    lowercase=True,
    stop_words="english",
    ngram_range=(1, 2),
    max_features=50000
)

retriever_matrix = retriever_vectorizer.fit_transform(
    retrieval_corpus["text_customer"]
)


def retrieve_similar_cases(query, top_k=3):
    query_vector = retriever_vectorizer.transform([query])

    similarities = cosine_similarity(
        query_vector,
        retriever_matrix
    ).flatten()

    top_indices = similarities.argsort()[-top_k:][::-1]

    results = retrieval_corpus.iloc[top_indices].copy()

    results["similarity"] = similarities[top_indices]

    return results


print("Retriever ready!")
```

```python
import os
from dotenv import load_dotenv
from groq import Groq

load_dotenv()

client = Groq(
    api_key=os.getenv("GROQ_API_KEY")
)

MODEL = "openai/gpt-oss-20b"


def generate_grounded_response(customer_message: str, retrieved_cases) -> str:

    evidence = ""

    for i, (_, row) in enumerate(retrieved_cases.iterrows(), start=1):
        evidence += f"""
Case {i}:
Customer: {row["text_customer"]}
Historical Uber response: {row["text_uber"]}
Similarity: {row["similarity"]:.3f}
"""

    prompt = f"""
You are an Uber customer support assistant.

Your task is to draft a helpful support response to the customer using
ONLY the information supported by the historical cases provided below.

Customer message:
"{customer_message}"

Historical cases:
{evidence}

Rules:
- Use the historical cases as evidence of how similar issues were handled.
- Do not invent Uber policies, refunds, credits, compensation, guarantees,
  investigations, or actions that are not supported by the evidence.
- Do not claim that you personally performed an investigation or took an action.
- If the historical evidence does not provide enough information to resolve
  the customer's issue, acknowledge the issue and direct the customer to
  contact Uber Support through the appropriate support channel.
- Do not mention the historical cases or similarity scores.
- Keep the response concise, professional, and empathetic.
- Respond directly to the customer.

Return ONLY the support response.
"""

    response = client.chat.completions.create(
        model=MODEL,
        messages=[
            {
                "role": "user",
                "content": prompt
            }
        ],
        max_completion_tokens=600,
        reasoning_effort="low",
        include_reasoning=False,
        temperature=0.2
    )

    content = response.choices[0].message.content

    if not content:
        return "Unable to generate a response."

    return content.strip()


print("Response generator ready!")
```

```python
import json
import re


def llm_judge_response(
    customer_message: str,
    generated_response: str,
    retrieved_cases
):
    # Build the evidence shown to the judge
    evidence = ""

    for i, (_, row) in enumerate(
        retrieved_cases.iterrows(),
        start=1
    ):
        evidence += f"""
Case {i}:
Customer: {row["text_customer"]}
Historical Uber response: {row["text_uber"]}
"""

    prompt = f"""
You are evaluating an AI-generated Uber customer support response.

You must score the GENERATED RESPONSE using the CUSTOMER MESSAGE
and RETRIEVED EVIDENCE below.

CUSTOMER MESSAGE:
{customer_message}

GENERATED RESPONSE:
{generated_response}

RETRIEVED EVIDENCE:
{evidence}

Score the generated response on four dimensions.

HELPFULNESS:
1 = Does not meaningfully address the customer's problem
2 = Partially addresses the problem
3 = Clearly addresses the problem

GROUNDING:
1 = Makes important claims or promises unsupported by the retrieved evidence
2 = Mostly grounded but contains some questionable assumptions
3 = Fully supported by the retrieved evidence

ACTIONABILITY:
1 = Provides no useful next step
2 = Provides a generic next step
3 = Provides a clear and useful next step

APPROPRIATENESS:
1 = Inappropriate, unsafe, or unsuitable
2 = Acceptable but weak
3 = Professional, appropriate, and suitable

Important:
- Judge the GENERATED RESPONSE, not the historical responses.
- A generated response can be better written or more helpful than the
  historical responses and still receive a grounding score of 3.
- For grounding, determine whether the claims and actions in the generated
  response are supported by the retrieved evidence.
- Do not penalize a response merely because the historical response is
  generic.
- Do not reward a response merely because it sounds professional.
- Return exactly one JSON object.
- Do not include explanations outside the JSON.

Return:
{{
  "helpfulness": 1,
  "grounding": 1,
  "actionability": 1,
  "appropriateness": 1
}}
"""

    response = client.chat.completions.create(
        model=MODEL,
        messages=[
            {
                "role": "user",
                "content": prompt
            }
        ],
        max_completion_tokens=600,
        reasoning_effort="low",
        include_reasoning=False,
        temperature=0
    )

    content = response.choices[0].message.content

    if not content:
        raise ValueError("LLM judge returned an empty response.")

    # Extract JSON in case the model adds surrounding text
    match = re.search(r"\{.*\}", content, re.DOTALL)

    if not match:
        raise ValueError(
            f"Could not find JSON in judge response:\n{content}"
        )

    scores = json.loads(match.group())

    # Validate scores
    dimensions = [
        "helpfulness",
        "grounding",
        "actionability",
        "appropriateness"
    ]

    for dimension in dimensions:
        if dimension not in scores:
            raise ValueError(
                f"Missing score: {dimension}"
            )

        if scores[dimension] not in [1, 2, 3]:
            raise ValueError(
                f"Invalid {dimension} score: {scores[dimension]}"
            )

    return scores


print("LLM judge ready!")
```
